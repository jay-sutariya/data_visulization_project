```python
dfcorr = df[['years_experience','salary','job_posting_month','job_openings']]
dfcorr.corr()
sns.heatmap(dfcorr.corr(),annot=True)

corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
